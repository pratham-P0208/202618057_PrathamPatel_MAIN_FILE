```python
import numpy as np
A1 = np.array([
    [1,2],
    [3,4]
])

A2 = np.array([
    [5,6],
    [7,8]
])
print(A1)
print("\n",A2)
add = A1 + A2
print("\nAddition: ",add)

ele_mul = A1*A2
print("\nElement-wise multiplication: ",ele_mul)

matrix_mul = A1 @ A2
print("\nMatrix multiplication: ",matrix_mul)
```

```python
mat = np.array([
    [1,2],
    [3,4]
])
print(mat)
transpose = mat.T
print("Transpose of MATRIX is:","\n",transpose)

det_mat= round(np.linalg.det(mat))
print("Determinant is:","\n",det_mat)
```

```python
mat2 = np.array([
    [5, 20],
    [33, 45]
])

mat_inv = np.linalg.inv(mat2)
print("Inverse matrix is: ","\n",mat_inv)

result = mat2 @ mat_inv

print("\n")
print(result)

verification = np.allclose(result, np.eye(2))
print("\nVerification:", verification)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("train.csv")
missing_cnt = df.isnull().sum()

missing_percentage = (
    df.isnull().sum() / len(df)
) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_cnt,
    "Missing Percentage": missing_percentage
})

print("Missing Value Summary:")
print(missing_summary)
```

```python
missing_cnt.plot(kind="bar")

plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.title("Missing Values by Column")

plt.show()
```

```python
print("Missing Age before:",
      df["Age"].isnull().sum())

df_mean = df.copy()

mean_age = df_mean["Age"].mean()

df_mean["Age"] = df_mean["Age"].fillna(mean_age)

print("Mean Age:", mean_age)

print("Missing Age after:",
      df_mean["Age"].isnull().sum())
```

```python
df_mean = df.copy()
mean_age = df_mean["Age"].mean()
df_mean["Age"] = df_mean["Age"].fillna(mean_age)

df_mode = df.copy()
mode_age = df_mode["Age"].mode()[0]
df_mode["Age"] = df_mode["Age"].fillna(mode_age)

df_median = df.copy()
median_age = df_median["Age"].median()
df_median["Age"] = df_median["Age"].fillna(median_age)

df_ran = df.copy()
np.random.seed(42)
miss_age = df_ran["Age"].isnull()
ran_age = np.random.choice(
    df_ran["Age"].dropna(),
    size=miss_age.sum()
)
df_ran.loc[miss_age, "Age"] = ran_age
print("\nMissing Age after imputation:")

print("Mean:",
      df_mean["Age"].isnull().sum())

print("Median:",
      df_median["Age"].isnull().sum())

print("Mode:",
      df_mode["Age"].isnull().sum())

print("Random:",
      df_ran["Age"].isnull().sum())
```

```python
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]
print("\nFare Outlier Analysis:")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
```

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)
print(df[["SibSp", "Parch", "FamilySize"]].head())
print("\nIsAlone:")
print(df[["FamilySize", "IsAlone"]].head(10))
```

```python
pivot = pd.pivot_table(
    df,index="Sex",columns="Pclass",values="Survived",aggfunc="mean")

print("Pivot Table:")
print(pivot)

highest_group = pivot.stack().idxmax()
highest_value = pivot.stack().max()

lowest_group = pivot.stack().idxmin()
lowest_value = pivot.stack().min()

print("\nHighest survival group:", highest_group)
print("Highest survival rate:", highest_value)

print("\nLowest survival group:", lowest_group)
print("Lowest survival rate:", lowest_value)
```
